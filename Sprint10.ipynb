## 1. Consulta l’API pública JSONPlaceholder utilitzant el mètode GET per obtenir:

Llista de publicacions (/posts)
Llista d’usuaris (/users)
Llista de tasques (/todos)

```python
import requests

base_url = "https://jsonplaceholder.typicode.com"

# Publicaciones

response_posts = requests.get(f"{base_url}/posts")
response_posts.raise_for_status()
posts = response_posts.json()

# Usuarios

response_users = requests.get(f"{base_url}/users")
response_users.raise_for_status()
users = response_users.json()

# Tareas

response_todos = requests.get(f"{base_url}/todos")
response_todos.raise_for_status()
todos = response_todos.json()

print("Publicaciones:", len(posts))
print("Usuarios:", len(users))
print("Tareas:", len(todos))
```

## 2. Mostra per pantalla:

La quantitat total de cada recurs.
El codi d'estat de cada petició.

```python
import requests

# URL base de la API
base_url = "https://jsonplaceholder.typicode.com"

# 1. Obtener publicaciones

response_posts = requests.get(
    f"{base_url}/posts"
)

posts = response_posts.json()


# 2. Obtener usuarios

response_users = requests.get(
    f"{base_url}/users"
)

users = response_users.json()


# 3. Obtener tareas

response_todos = requests.get(
    f"{base_url}/todos"
)

todos = response_todos.json()


# 4. Mostrar cantidad total y código de estado

print("PUBLICACIONES")
print("Cantidad total:", len(posts))
print("Código de estado:", response_posts.status_code)

print("\nUSUARIOS")
print("Cantidad total:", len(users))
print("Código de estado:", response_users.status_code)

print("\nTAREAS")
print("Cantidad total:", len(todos))
print("Código de estado:", response_todos.status_code)
```

##4. Fes una petició POST per crear una nova publicació fictícia. Inclou el títol, el cos del missatge i un userId.

Mostra la resposta JSON.
Mostra el codi d'estat.

```python

# 6. Crear una nueva publicación mediante POST

new_post = {
    "title": "Nueva publicación ficticia",
    "body": "Este es el contenido de la nueva publicación.",
    "userId": 1
}

response_post = requests.post(
    f"{base_url}/posts",
    json=new_post
)

print("\nNUEVA PUBLICACIÓN")
print("Respuesta JSON:")
print(response_post.json())

print("Código de estado:", response_post.status_code)
```

## 5. Fes una petició PATCH per modificar parcialment una publicació existent.

Mostra la resposta JSON.
Mostra el codi d'estat.

```python

# 7. Modificar parcialmente una publicación mediante PATCH

update_post = {
    "title": "Título modificado"
}

response_patch = requests.patch(
    f"{base_url}/posts/1",
    json=update_post
)

print("\nPUBLICACIÓN MODIFICADA")
print("Respuesta JSON:")
print(response_patch.json())

print("Código de estado:", response_patch.status_code)
```

##1. Utilitza l’API d’Open Data BCN per:

Cercar un dataset del teu interès mitjançant package_search i package_show.

```python
import requests

url = "https://opendata-ajuntament.barcelona.cat/data/api/action/package_search"

params = {
    "q": "Bicing"
}

response = requests.get(url, params=params)

print("Código de estado:", response.status_code)

data = response.json()

print("Número de datasets encontrados:", data["result"]["count"])

for dataset in data["result"]["results"]:
    print(dataset["name"])
    print(dataset["title"])
    print()
```

```python
dataset_name = data["result"]["results"][3]["name"]

url = "https://opendata-ajuntament.barcelona.cat/data/api/action/package_show"

params = {
    "id": dataset_name
}

response = requests.get(url, params=params)

print("Código de estado:", response.status_code)

dataset = response.json()

print("Título:", dataset["result"]["title"])
print("Descripción:", dataset["result"]["notes"])
```

##2. Dels resultats obtinguts, selecciona’n un que tingui recursos disponibles en CSV o JSON.

```python
import requests
# Buscamos recursos con disponibilidad en el datastore

url = "https://opendata-ajuntament.barcelona.cat/data/api/action/package_search"

params = {
    "q": "Bicing",
    "rows": 20
}

response = requests.get(url, params=params)

search_data = response.json()

for ds in search_data["result"]["results"]:
    for recurso in ds.get("resources", []):
        if recurso.get("datastore_active") == True:
            print("Dataset:", ds["title"])
            print("Recurso:", recurso["name"])
            print("Formato:", recurso["format"])
            print("Resource ID:", recurso["id"])
            print()

# Selecciono uno y muestro los 5 primeros registros

resource_id = "8cd14603-4a48-4425-9683-4ac5b94faea6"

url = "https://opendata-ajuntament.barcelona.cat/data/api/action/datastore_search"

params = {
    "resource_id": resource_id,
    "limit": 100
}

response = requests.get(url, params=params)

print("Código de estado:", response.status_code)

data_store = response.json()

print("Registros recuperados:", len(data_store["result"]["records"]))

print("\nPrimeros 5 registros:")
for registro in data_store["result"]["records"][:5]:
    print(registro)
```

##3. Amb el resource_id del recurs seleccionat, realitza una consulta mitjançant datastore_search per recuperar almenys 100 registres.

```python
import requests

resource_id = "8cd14603-4a48-4425-9683-4ac5b94faea6"

url = "https://opendata-ajuntament.barcelona.cat/data/api/action/datastore_search"

params = {
    "resource_id": resource_id,
    "limit": 100
}

response = requests.get(url, params=params)

print("Código de estado:", response.status_code)

data = response.json()

print("Registros recuperados:", len(data["result"]["records"]))

print("\nPrimeros 100 registros:")
for registro in data["result"]["records"][:100]:
    print(registro)
```
