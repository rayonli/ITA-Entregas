####1.-Recoger inscripciones: Definimos tipos de dato y creamos una lista para almacenar los datos introducidos por los candidatos. También validamos que el tipo de dato sea correcto.

```python
def obtener_datos():

  try:
      nombre = str(input("Nombre y apellidos: ").strip())
      telefono = str(input("Teléfono: "))
      experiencia = int(input("Años de experiencia: "))
      catalan = int(input("Catalán (0-3): "))
      excel = int(input("Excel (0-3): "))
      titulacion = int(input("Titulación (0-3): "))

      nuevo_candidato = [
            nombre,
            telefono,
            experiencia,
            catalan,
            excel,
            titulacion
        ]

      return nuevo_candidato

  except ValueError:
      print("Hay un error en los datos. Introdúcelos de nuevo.")
```

####2.-Validar datos: controlamos lo que el usuario introduce en el formulario.

```python
def validar_datos(nuevo_candidato):

    nuevo_candidato[0] = nuevo_candidato[0].strip()
    valido = True

    if (nuevo_candidato[0] == ""):
          print("El nombre no puede estar vacío.")
          return False
    if (len(nuevo_candidato[1]) != 9):
          print("El teléfono ha de tener 9 dígitos")
          return False
    if (not nuevo_candidato[1].isdigit()):
          print("El número de teléfono solo puede contener dígitos")
          return False
    if (nuevo_candidato[2] > 60 or nuevo_candidato[2] < 0):
          print("Revisa tus años de experiencia.")
          return False
    if (nuevo_candidato[3] > 3 or nuevo_candidato[3] < 0):
          print("Revisa tu nivel de catalán.")
          return False
    if (nuevo_candidato[4] > 3 or nuevo_candidato[4] < 0):
          print("Revisa tu nivel de excel.")
          return False
    if (nuevo_candidato[5] > 3 or nuevo_candidato[5] < 0):
        print("Revisa tu nivel de estudios.")
        return False
    return True

```

####3.-Calcular puntuación.

```python
def calcular_puntuacion(nuevo_candidato):
  puntuacion = (nuevo_candidato[2] * 1.5) + nuevo_candidato[3] + nuevo_candidato[4] + nuevo_candidato[5]
  return puntuacion
```

####4.-Mostrar datos individuales: de cada candidato.

```python
def mostrar_individual(nuevo_candidato):
  print("Se acepta la candidatura de", nuevo_candidato[0], "con teléfono", nuevo_candidato[1], ", que cuenta con una puntuación de", nuevo_candidato[6], ".")
```

####5.-Clasificar candidato: asignamos prioridad.

```python
def clasificar_candidato(nuevo_candidato):
  if nuevo_candidato[6] >= 12:
    return "Prioridad Alta."
  elif nuevo_candidato[6] >= 8:
    return "Candidato adecuado."
  else:
    return "No prioritario."
```

####6.-Crear informe final.

```python
def mostrar_informe(candidatos):

    prioridad = {
        "Prioridad Alta.": 1,
        "Candidato adecuado.": 2,
        "No prioritario.": 3
    }

    candidatos.sort(
        key=lambda candidato: (
            prioridad[candidato[7]],
            -candidato[6]
        )
    )

    print("\nINFORME FINAL")
    print("-" * 60)

    for candidato in candidatos:
        print(
            "Nombre:", candidato[0],
            "Teléfono:", candidato[1],
            "Puntuación:", candidato[6],
            "Clasificación:", candidato[7]
        )
```

####7.-Main seleccion: Función que orquesta las demás: creamos una lista para almacenar a los candidatos hasta que tengamos 10, pero primero almacenamos el candidato en la variable "posible_candidato", lo pasamos por la función de validación, y si es correcto, entonces calculamos la puntuación, la añadimos a "posible_candidato", le asignamos una prioridad que almacenamos en la variable "clasificación", y la añadimos como item a la lista "posible_candidato". Mostramos los datos del candidato, y lo añadimos a la lista "candidatos", para mostrar el informe final.

```python
def main_seleccion():
  candidatos = []

  while len(candidatos) < 10:

    posible_candidato = obtener_datos()

    if validar_datos(posible_candidato) == True:
      puntuacion = calcular_puntuacion(posible_candidato)

      posible_candidato.append(puntuacion)

      clasificacion = clasificar_candidato(posible_candidato)

      posible_candidato.append(clasificacion)

      mostrar_individual(posible_candidato)

      candidatos.append(posible_candidato)

  mostrar_informe(candidatos)


```

###Ejecutamos programa.

```python
main_seleccion()
```

Disposes d'un conjunt de dades amb informació bàsica de 10 clients:

El departament de CRM ha definit manualment:

» Codis de comportament
0 → Navega però no afegeix productes
1 → Afegeix productes però no arriba al checkout
2 → Arriba al checkout però no paga
3 → Cistella de compra abandonat fa menys de 24 h
4 → Cistella de compra abandonat fa més de 24 h

» Factors segons codi de comportament
Codi 0 → 0.10
Codi 1 → 0.25
Codi 2 → 0.50
Codi 3 → 0.75
Codi 4 → 0.90

» Factors segons grup d'edat
18–25 → 1.20
26–40 → 1.00
41–60 → 0.85
60+ → 0.70

» Fórmula del risc final
risc_final = factor codi * factor edat
Un cop processats els clients, has de ser capaç de respondre:

» 1. Quins clients s'han de contactar primer en una campanya de retargeting?
» 2. Quines diferències hi ha entre clients amb el mateix comportament però d'edats diferents? (Per exemple: un client jove que abandona la cistella de compra fa 24 h té el mateix risc que un de més edat?)
» 3. Quins clients presenten tan poc interès que no val la pena invertir-hi pressupost?
» 4. Quin client seria el més rendible de recuperar?


###Clasificar segmento y factor -> clasificar edad -> calcular riesgo -> procesar clientes -> mostrar informe

####Definición de variables globales

####1.-Clasificar segmento y factor: creamos un diccionario "codigos" con los mensajes y factores asignados.

```python
def obtener_segmento_y_factor(codigo_comportamiento):
  codigos = {

    0:{
        "mensaje":"Navega pero no añade productos",
        "factor":0.10
    },

    1:{
        "mensaje":"Añade productos pero no llega al checkout",
        "factor":0.25
    },

    2:{
        "mensaje":"Llega al checkout pero no paga",
        "factor":0.50
    },

    3:{
        "mensaje":"Cesta de la compra abandonada hace menos de 24 horas",
        "factor":0.75
    },

    4:{
        "mensaje":"Cesta de la compra abandonada hace más de 24 horas",
        "factor":0.90
    }

}
  return codigos[codigo_comportamiento]
```

####2.-Obtener factor por edad: realizamso el cálculo para obtener el factor según la edad.

```python
def obtener_factor_edad(edad):

    if 18 <= edad <= 25:
        return "18-25",1.20

    elif 26 <= edad <= 40:
        return "26-40",1.00

    elif 41 <= edad <= 60:
        return "41-60",0.85

    else:
        return "60+",0.70
```

####3.-Calcular riesgo

```python
def calcular_riesgo(factor_codigo,factor_edad):

    calculo = factor_codigo * factor_edad
    return round(calculo,2)
```

####4.-Procesar clientes: almacena el mensaje y factor en la variable "info", y la edad yel factor en otras 2 variables, que se envían a la función "calcular riesgo", para finalmente crear un diccionario "cliente final" en el que ya incluye todos los parámetros.

```python
def procesar_clientes(cliente):

    info = obtener_segmento_y_factor(cliente["codigo_comportamiento"])

    grupo_edad, factor_edad = obtener_factor_edad(cliente["edad"])

    riesgo = calcular_riesgo(
        info["factor"],
        factor_edad
    )

    cliente_final = {

        "id_cliente":cliente["id"],
        "nombre":cliente["nombre"],
        "edad":cliente["edad"],
        "grupo_edad":grupo_edad,
        "codigo_comportamiento":cliente["codigo_comportamiento"],
        "mensaje":info["mensaje"],
        "factor_codigo":info["factor"],
        "factor_edad":factor_edad,
        "riesgo_final":riesgo

    }

    return cliente_final
```

####5.-Mostrar dataframe

```python
def mostrar_dataframe(clientes):

    print("\nCLIENTES")

    for cliente in clientes:

        print("----------------------------")

        for clave,valor in cliente.items():

            print(clave,":",valor)
```

####6.-Mostrar respuestas: Muestra los clientes prioritarios ordenándolos por "riesgo_final" de mayor a menor. Muestra las diferencias por edad mostrando el factor de la edad. Muestra los clientes poco interesantes según el baremos de "riesgo_final" dado. Y muestra el cliente más rentable al mostrar el primero del diccionario que ordenamos por "riesgo_final".

```python
def mostrar_respuestas(clientes):

    print("\n1. Clientes prioritarios")

    clientes.sort(
        key=lambda x:x["riesgo_final"],
        reverse=True
    )

    for cliente in clientes:

        print(
            cliente["nombre"],
            round(cliente["riesgo_final"],2)
        )

    print("\n2. Diferencias por edad")

    for cliente in clientes:

        print(
            cliente["nombre"],
            cliente["grupo_edad"],
            cliente["factor_edad"]
        )

    print("\n3. Poco interesantes")

    for cliente in clientes:

        if cliente["riesgo_final"] <= 0.25:

            print(cliente["nombre"])

    print("\n4. Cliente más rentable")

    mejor = clientes[0]

    print(
        mejor["nombre"],
        mejor["riesgo_final"]
    )
```

###Limpiar lista pacientes -> calcular riesgo -> ordenar por riesgo -> mostrar informe de pacientes con riesgo total e informe ordenado por prioridad

####Definición de variables globales

```python
pacientes = [
    "Maria|45|tos, Fiebre Alta, dolor_pecho",
    "Luis|33|dolor_cabeza,faTiga",
    "Sara|67|fiebre alta, dificultad RESPIRAR, tos",
    "Jordi|52|fatiga , tos",
    "Anna|29|fiebre alta,dolor_cabeza"
]
niveles_riesgo = {
    "fiebre_alta": 3,
    "dificultad_respirar": 5,
    "dolor_pecho": 4,
    "tos": 1,
    "fatiga": 1,
    "dolor_cabeza": 1
  }
```

####1.-Limpiar lista pacientes: separamos nombre, edad y síntomas. Eliminamos espacios al principio o final, transformamos a minúsculas y reemplazamos los espacios por guiones bajos. Por último, añadimos los pacientes normalizados a un diccionario.

```python
def limpiar_lista_pacientes(pacientes):

    resultado = []

    for paciente in pacientes:

        datos = paciente.split("|")

        sintomas = datos[2].split(",")

        lista_sintomas = []

        for sintoma in sintomas:
            lista_sintomas.append(
                sintoma.strip().lower().replace(" ", "_")
            )

        paciente_dic = {
            "nombre": datos[0],
            "edad": int(datos[1]),
            "sintomas": lista_sintomas
        }

        resultado.append(paciente_dic)

    return resultado
```

####2.-Calcular riesgo

```python
def calcular_riesgo(dic_pacientes, dic_niveles_riesgo):
  for paciente in dic_pacientes:
   riesgo_total = 0
   for sintoma in paciente["sintomas"]:
     riesgo_total += dic_niveles_riesgo[sintoma]
   paciente["riesgo_total"] = riesgo_total
  return dic_pacientes
```

####3.-Ordenar pacientes

```python
def ordenar_pacientes(dic_pacientes):
    dic_pacientes.sort(
        key=lambda paciente: paciente["riesgo_total"],
        reverse=True
    )
    return dic_pacientes
```

####4.-Mostrar prioridad

```python
def mostrar_prioridad(dic_pacientes):

    print("\nPACIENTES ORDENADOS POR RIESGO")
    print("Nombre\tEdad\tRiesgo\tSíntomas")

    for paciente in dic_pacientes:

        print(
            paciente["nombre"],
            "\t",
            paciente["edad"],
            "\t",
            paciente["riesgo_total"],
            "\t",
            paciente["sintomas"]
        )
```

####5.-Main risk: Función que orquesta las demás: enviamos la lista de pacientes a la función limpiar, y el resultado lo almacenamos en "pacientes_procesados", que volvemos a enviar junto con los niveles dados a la función "calcular_riesgo", y el resultado lo almacenamos en "diccionario_pacientes_riesgo". Iteramos en él para mostrar el nombre, edad, riesgo_total y síntomas. Lo enviamos también a la función "ordenar_pacientes", y el resultado lo almacenamos en "pacientes_ordenados", que utilizamos para mostrar la prioridad.

```python
def main_risk(lista_pacientes):

  pacientes_procesados = limpiar_lista_pacientes(lista_pacientes)
  diccionario_pacientes_riesgo = calcular_riesgo(pacientes_procesados,niveles_riesgo)
  print("Nombre\tEdad\tRiesgo\tSíntomas")
  for paciente in diccionario_pacientes_riesgo:
      print(
        paciente["nombre"],
        "\t",
        paciente["edad"],
        "\t",
        paciente["riesgo_total"],
        "\t",
        paciente["sintomas"]
      )
  pacientes_ordenados = ordenar_pacientes(diccionario_pacientes_riesgo)
  mostrar_prioridad(pacientes_ordenados)

```

###Ejecutamos programa.

```python
main_risk(pacientes)
```

###procesar_seguro -> mostrar_clientes -> ordenar_clientes -> mostrar_clientes_ordenados -> mostrar_analisis

####Definición de variables externas

```python


clientes = [
    {"nombre": "Ana", "edad": 42, "código": 0},
    {"nombre": "Carlos", "edad": 30, "código": 2},
    {"nombre": "Isabel", "edad": 55, "código": 3},
    {"nombre": "Jorge", "edad": 40, "código": 1},
    {"nombre": "Marta", "edad": 28, "código": 0}
]

niveles_seguro = {
    0: {"categoría": "Bajo riesgo", "precio_base": 120, "fraude": 0.01},
    1: {"categoría": "Riesgo medio", "precio_base": 200, "fraude": 0.05},
    2: {"categoría": "Riesgo alto", "precio_base": 350, "fraude": 0.15},
    3: {"categoría": "Riesgo crítico", "precio_base": 500, "fraude": 0.30}
}
```

####1.-Cliente final: recibe lista de clientes y lista de niveles de seguro y devuelve una lista de clientes a la que se ha añadido probabilidad de fraude y precio final.

```python
def clientes_final(lista_clientes_procesados, lista_niveles_seguro):

    lista_clientes_seguro = []

    for cliente in lista_clientes_procesados:

        info = lista_niveles_seguro[cliente["código"]]

        precio_final = info["precio_base"] * (1 + cliente["edad"] / 100)

        cliente_final = {

            "nombre": cliente["nombre"],
            "edad": cliente["edad"],
            "categoría": info["categoría"],
            "precio_base": info["precio_base"],
            "precio_final": precio_final,
            "probabilidad_fraude": info["fraude"]

        }

        lista_clientes_seguro.append(cliente_final)

    return lista_clientes_seguro
```

####2.-Mostrar clientes: muestra la lista creada.

```python
def mostrar_clientes_final(lista_clientes):

    print("\nCLIENTES")
    print("Nombre\tEdad\tCategoría\tPrecio Base\tPrecio Final   Fraude")

    for cliente in lista_clientes:

        print("{:<8} {:<5} {:<18} {:<17} {:<11} {:<8}".format(
            cliente["nombre"],
            cliente["edad"],
            cliente["categoría"],
            cliente["precio_base"],
            round(cliente["precio_final"],2),
            cliente["probabilidad_fraude"]
        ))
```

####3.-Ordenar clientes: ordena por probabilidad de fraude y, en caso de empate, por precio final.

```python
def ordenar_clientes(lista_clientes):

    lista_clientes.sort(

        key=lambda cliente: (

            -cliente["probabilidad_fraude"],
            -cliente["precio_final"]

        )

    )

    return lista_clientes
```

####4.-Mostrar clientes ordenados.

```python
def mostrar_clientes_ordenados(lista_clientes):

    print("\nCLIENTES ORDENADOS")

    print("Nombre\tEdad\tCategoría\tPrecio_Final   Fraude")

    for cliente in lista_clientes:

        print("{:<8} {:<5} {:<18} {:>5} {:>12}".format(

            cliente["nombre"],
            cliente["edad"],
            cliente["categoría"],
            round(cliente["precio_final"],2),
            cliente["probabilidad_fraude"]

        ))
```

####5.-Mostrar análisis: muestra clientes con mayor riesgo de fraude, con mayor precio final de seguro, y el prioritario para revisar manualmente.

```python
def mostrar_analisis(lista_clientes_analizar):
  mayor_fraude = lista_clientes_analizar[0]
  print("\n""Mayor riesgo de fraude:",mayor_fraude["nombre"])
  mayor_precio = max(
    lista_clientes_analizar,
    key=lambda cliente:cliente["precio_final"]
)
  print(
    "\nPrecio más alto:",
    mayor_precio["nombre"],
    mayor_precio["precio_final"],
    "€"
)
  print(
    "\nPrioritario para revisar manualmente:",
    lista_clientes_analizar[0]["nombre"]
)
```

####6.-Main seguro: Función que orquesta las demás: almacenamos en "cliente_seguro" el diccionario que devuelve la función "cliente_final", con precio_final y probabilidad de fraude añadidos. Lo mostramos enviándolo a la función "mostrar_clientes_final". Lo almacenamos ordenado en "clientes_ordenados" tras enviarlo a la función "ordenar_clientes", para finalmente mostrarlos ordenados, y mostrar el análisis, invocando a ambas funciones.

```python
def main_seguro(lista_clientes):

    clientes_seguro = clientes_final(
        lista_clientes,
        niveles_seguro
    )

    mostrar_clientes_final(clientes_seguro)

    clientes_ordenados = ordenar_clientes(clientes_seguro)

    mostrar_clientes_ordenados(clientes_ordenados)

    mostrar_analisis(clientes_ordenados)
```

###Crear caracteres -> Generar contraseña

####Importamos librerías para usar el método random y cadenas de texto y caracteres

```python
import numpy as np
import string
```

####1.-Crear caracteres: función que genera caracteres mayúsculas, minúsculas, números y signos, que se almacenan en una lista "disponibles", y de la que, aleatoriamente se añaden a otra lista " obligatorios".

```python
def crear_caracteres(mayusculas, minusculas, numeros, signos):

    disponibles = []
    obligatorios = []

    if mayusculas:
        disponibles += list(string.ascii_uppercase)
        obligatorios.append(
            np.random.choice(list(string.ascii_uppercase))
        )

    if minusculas:
        disponibles += list(string.ascii_lowercase)
        obligatorios.append(
            np.random.choice(list(string.ascii_lowercase))
        )

    if numeros:
        disponibles += list(string.digits)
        obligatorios.append(
            np.random.choice(list(string.digits))
        )

    if signos:
        disponibles += list(string.punctuation)
        obligatorios.append(
            np.random.choice(list(string.punctuation))
        )

    return disponibles, obligatorios
```

####2.-Función que genere contraseñas: captura las listas "disponibles" y "obligatorios", crea una copia de "obligatorios" en "contraseña", y mientras la longitud sea inferior a la deseada, irá añadiendo caracteres de "disponibles". Luego los mezcla, y los une, ya que inicialmente son un listado de items string.

```python
def generar_contraseña(
        longitud,
        mayusculas=True,
        minusculas=True,
        numeros=True,
        signos=False
):

    disponibles, obligatorios = crear_caracteres(
        mayusculas,
        minusculas,
        numeros,
        signos
    )

    contraseña = obligatorios.copy()

    while len(contraseña) < longitud:

        contraseña.append(
            np.random.choice(disponibles)
        )

    np.random.shuffle(contraseña)

    return "".join(contraseña)
```

####3.-Main: Función que orquesta las demás. Aquí lo más significativo es el control de que no se pueden anular todos a la vez, puesto que devuelve error. Al menos un grupo de caracteres ha de ser True.

```python
def main():

    mayusculas = True
    minusculas = True
    numeros = True
    signos = True

    if (mayusculas == False and minusculas == False and numeros == False and signos == False):

        print("No has seleccionado ningún grupo de caracteres.")

    else:

        password = generar_contraseña(
            longitud=12,
            mayusculas=mayusculas,
            minusculas=minusculas,
            numeros=numeros,
            signos=signos
        )

        print("Contraseña generada:")
        print(password)
```

###Ejecutamos programa.

```python
main()
```

###Leer fichero -> Procesar partidos (Crear equipo, actualizar goles, actualizar puntos) -> Mostrar tabla -> Obtener estadísticas -> Ordenar clasificación -> Mostrar clasificación

####1.-Leer fichero.

```python
def leer_fichero():

    fichero = open("historic partits.txt","r",encoding="utf-8")

    partidos = fichero.readlines()

    fichero.close()

    return partidos
```

####2.-Procesar partidos: Extraemos los datos de local y visitante, de resultados y goles, y los separamos por equipos, actualizando los goles y los puntos, y devolviendo un diccionario.

```python
def procesar_partidos(partidos):

    equipos = {}

    for partido in partidos:

        datos = partido.strip().split("\t")

        local = datos[0]

        visitante = datos[2]

        resultado = datos[1].split("-")

        goles_local = int(resultado[0])

        goles_visitante = int(resultado[1])

        crear_equipo(equipos, local)

        crear_equipo(equipos, visitante)

        actualizar_goles(
            equipos,
            local,
            visitante,
            goles_local,
            goles_visitante
        )

        actualizar_puntos(
            equipos,
            local,
            visitante,
            goles_local,
            goles_visitante
        )

    return equipos
```

#####2.1.-Crear equipo.

```python
def crear_equipo(equipos, nombre):

    if nombre not in equipos:

        equipos[nombre] = {
    "goles_favor": 0,
    "goles_contra": 0,
    "puntos": 0,
    "ganados": 0,
    "empatados": 0,
    "perdidos": 0
}
```

#####2.2.-Actualizar goles.

```python
def actualizar_goles(
    equipos,
    local,
    visitante,
    goles_local,
    goles_visitante
):

    equipos[local]["goles_favor"] += goles_local

    equipos[local]["goles_contra"] += goles_visitante

    equipos[visitante]["goles_favor"] += goles_visitante

    equipos[visitante]["goles_contra"] += goles_local
```

#####2.3.-Actualizar puntos.

```python
def actualizar_puntos(equipos, local, visitante, goles_local, goles_visitante):

    if goles_local > goles_visitante:

        equipos[local]["puntos"] += 3
        equipos[local]["ganados"] += 1

        equipos[visitante]["perdidos"] += 1

    elif goles_local < goles_visitante:

        equipos[visitante]["puntos"] += 3
        equipos[visitante]["ganados"] += 1

        equipos[local]["perdidos"] += 1

    else:

        equipos[local]["puntos"] += 1
        equipos[visitante]["puntos"] += 1

        equipos[local]["empatados"] += 1
        equipos[visitante]["empatados"] += 1
```

####3.-Mostrar tabla.

```python
def mostrar_tabla(equipos):

    print("\nESTADÍSTICAS DE LOS EQUIPOS")
    print("{:<22} {:>4} {:>4} {:>4} {:>4} {:>4} {:>4}".format(
    "Equipo", "GF", "GC", "G", "E", "P", "Pts"
  ))

    for equipo in equipos:

      print("{:<22} {:>4} {:>4} {:>4} {:>4} {:>4} {:>4}".format(
        equipo,
        equipos[equipo]["goles_favor"],
        equipos[equipo]["goles_contra"],
        equipos[equipo]["ganados"],
        equipos[equipo]["empatados"],
        equipos[equipo]["perdidos"],
        equipos[equipo]["puntos"]
    ))
```

####4.-Mostrar estadísticas.

```python
def mostrar_estadisticas(equipos):

    goleador = max(

        equipos,

        key=lambda equipo:equipos[equipo]["goles_favor"]

    )

    goleado = max(

        equipos,

        key=lambda equipo:equipos[equipo]["goles_contra"]

    )

    print("\nEquipo más goleador:",goleador)

    print("Equipo más goleado:",goleado)
```

####5.-Mostrar clasificación.

```python
def mostrar_clasificacion(equipos):

    clasificacion = sorted(

        equipos.items(),

        key=lambda equipo:equipo[1]["puntos"],

        reverse=True

    )

    print("\nCLASIFICACIÓN")

    posicion = 1

    for equipo in clasificacion:

        print("{:<3} {:<21} {:<7}".format(

            posicion,

            equipo[0],

            equipo[1]["puntos"]

        ))

        posicion += 1
```

####6.-Main liga: Función que orquesta a las demás.

```python
def main_liga():

    partidos = leer_fichero()

    equipos = procesar_partidos(partidos)

    mostrar_tabla(equipos)

    mostrar_estadisticas(equipos)

    mostrar_clasificacion(equipos)
```

###Ejecutamos programa.

```python
main_liga()
```
